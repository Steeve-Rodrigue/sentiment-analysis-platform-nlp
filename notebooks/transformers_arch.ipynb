```python
import math

import torch
import torch.nn as nn

torch.manual_seed(42)

# Une phrase de 4 mots, deja "embeddee" (dimension 8)
phrase = torch.randn(1, 4, 8)
mots = ["the", "delivery", "was", "terrible"]

d_model = 8
d_k = 8  # dimension des projections Q, K, V (simplifie ici, egal a d_model)

W_q = nn.Linear(d_model, d_k, bias=False)
W_k = nn.Linear(d_model, d_k, bias=False)
W_v = nn.Linear(d_model, d_k, bias=False)

Q = W_q(phrase)
K = W_k(phrase)
V = W_v(phrase)

print("Formes de Q, K, V :", Q.shape, K.shape, V.shape)
print("-> chaque mot a maintenant 3 versions de lui-meme\n")

# Etape 1 : scores de compatibilite Q . K^T
scores = Q @ K.transpose(-2, -1)
print("Matrice de scores BRUTS (4x4, avant mise a l'echelle) :")
print(scores[0].detach().numpy().round(2))

# Etape 2 : mise a l'echelle
scores_scaled = scores / math.sqrt(d_k)
print("\nApres mise a l'echelle (/ sqrt(", d_k, ")) :")
print(scores_scaled[0].detach().numpy().round(2))
```

```python
import math

import torch
import torch.nn as nn

torch.manual_seed(42)

d_model = 8  # dimension totale
n_heads = 2  # 2 tetes en parallele
d_k = d_model // n_heads  # chaque tete travaille sur une portion plus petite

print(f"d_model={d_model}, n_heads={n_heads} -> chaque tete a d_k={d_k}")

phrase = torch.randn(1, 4, d_model)  # 4 mots

W_q = nn.Linear(d_model, d_model, bias=False)
W_k = nn.Linear(d_model, d_model, bias=False)
W_v = nn.Linear(d_model, d_model, bias=False)
W_o = nn.Linear(d_model, d_model, bias=False)  # projection finale

Q = W_q(phrase)  # (1, 4, 8)
K = W_k(phrase)
V = W_v(phrase)


# ETAPE CLE : decouper la dimension 8 en 2 tetes de 4
def decouper_en_tetes(x, n_heads, d_k):
    batch, longueur, _ = x.shape
    x = x.view(batch, longueur, n_heads, d_k)
    return x.permute(0, 2, 1, 3)  # (batch, n_heads, longueur, d_k)


Q_tetes = decouper_en_tetes(Q, n_heads, d_k)
K_tetes = decouper_en_tetes(K, n_heads, d_k)
V_tetes = decouper_en_tetes(V, n_heads, d_k)

print("Forme de Q apres decoupage en tetes :", Q_tetes.shape)
print("-> (batch, 2 tetes, 4 mots, 4 dimensions par tete)")
```

## 1.   Attention   mechanism

```python
# Prerequis : uv add torch
# Utilise les fonctions reelles de src/transformer_arch/attention_mechanism.py

# --- BLOC 1 : self-attention de base, matrice de poids ---
import os
import sys

sys.path.insert(0, os.path.abspath(os.path.join("..", "src")))
import torch

from transformers_arch.attention_mechanism import (
    MultiHeadAttention,
    SelfAttention,
    positional_encoding,
    scaled_dot_product_attention,
)

torch.manual_seed(42)
mots = ["the", "delivery", "was", "terrible"]
phrase = torch.randn(1, 4, 8)

model = SelfAttention(d_model=8)
sortie, poids = model(phrase)

print("Matrice des poids d'attention :\n")
print(" " * 12 + "".join(f"{m:>10}" for m in mots))
for i, ligne in enumerate(mots):
    valeurs = " ".join(f"{v:9.3f}" for v in poids[0, i].detach().numpy())
    print(f"{ligne:12}{valeurs}")

print("\nSomme de chaque ligne :", poids[0].sum(dim=-1).detach().numpy().round(4))
```

```python
# --- BLOC 2 : preuve -- le self-attention seul est invariant par ---
# --- permutation (aucune notion d'ordre) ---
torch.manual_seed(42)
model2 = SelfAttention(d_model=8)

mot_a = torch.randn(1, 1, 8)
mot_b = torch.randn(1, 1, 8)
mot_c = torch.randn(1, 1, 8)

phrase_1 = torch.cat([mot_a, mot_b, mot_c], dim=1)  # a b c
phrase_2 = torch.cat([mot_c, mot_b, mot_a], dim=1)  # c b a

sortie_1, _ = model2(phrase_1)
sortie_2, _ = model2(phrase_2)

print("\nSortie 'a' dans 'a b c' :", sortie_1[0, 0].detach().numpy().round(3))
print("Sortie 'a' dans 'c b a' :", sortie_2[0, 2].detach().numpy().round(3))
print("Identiques ?", torch.allclose(sortie_1[0, 0], sortie_2[0, 2], atol=1e-5))
print("-> confirme : sans position, l'ordre n'a aucune influence")
```

```python
# --- BLOC 3 : le positional encoding casse cette symetrie ---
pe = positional_encoding(max_len=3, d_model=8).unsqueeze(0)

sortie_1_pe, _ = model2(phrase_1 + pe)
sortie_2_pe, _ = model2(phrase_2 + pe)

print("\nAvec positional encoding ajoute :")
print("Identiques ?", torch.allclose(sortie_1_pe[0, 0], sortie_2_pe[0, 2], atol=1e-5))
print("-> maintenant different : le modele distingue la position")
```

```python
# --- BLOC 4 : masquage avec -inf, verifier le poids exact ---
q = torch.randn(1, 3, 4)
k = torch.randn(1, 3, 4)
v = torch.randn(1, 3, 4)

mask = torch.zeros(1, 3, 3, dtype=torch.bool)
mask[0, :, 2] = True  # masque toute la colonne 2 (ex: position de padding)

_, poids_masques = scaled_dot_product_attention(q, k, v, mask)
print("\nPoids avec la colonne 2 masquee :")
print(poids_masques[0].detach().numpy().round(3))
print("Somme de chaque ligne :", poids_masques[0].sum(dim=-1).numpy().round(4))
print("-> la colonne masquee est exactement 0, le reste se renormalise a 1")
```

```python
# --- BLOC 5 : multi-head -- tetes differentes, poids differents ---
torch.manual_seed(42)
mha = MultiHeadAttention(d_model=8, n_heads=2)
sortie_mha, poids_mha = mha(phrase)

print("\nForme des poids multi-head :", poids_mha.shape)
print("-> (batch, n_heads, longueur, longueur)")

tete_1 = poids_mha[0, 0]
tete_2 = poids_mha[0, 1]
print("\nTete 1 :\n", tete_1.detach().numpy().round(3))
print("Tete 2 :\n", tete_2.detach().numpy().round(3))
print("Identiques ?", torch.allclose(tete_1, tete_2, atol=1e-4))
```

## 2.   Encodeur

```python
# Prerequis : uv add torch
# Utilise les fonctions reelles de src/transformer_arch/encoder.py

# --- BLOC 1 : gradient avec vs sans connexions residuelles ---
import os
import sys

sys.path.insert(0, os.path.abspath(os.path.join("..", "src")))
import torch
from torch import nn

torch.manual_seed(42)


class BlocSansResiduel(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.couche = nn.Linear(dim, dim)

    def forward(self, x):
        return torch.relu(self.couche(x))


class BlocAvecResiduel(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.couche = nn.Linear(dim, dim)

    def forward(self, x):
        return x + torch.relu(self.couche(x))


dim, profondeur = 16, 30
reseau_sans = nn.Sequential(*[BlocSansResiduel(dim) for _ in range(profondeur)])
reseau_avec = nn.Sequential(*[BlocAvecResiduel(dim) for _ in range(profondeur)])

x = torch.randn(1, dim, requires_grad=True)
reseau_sans(x).sum().backward()
grad_sans = x.grad.abs().mean().item()

x2 = x.detach().clone().requires_grad_(True)
reseau_avec(x2).sum().backward()
grad_avec = x2.grad.abs().mean().item()

print(f"Gradient SANS residuel (30 couches) : {grad_sans:.8f}")
print(f"Gradient AVEC residuel (30 couches) : {grad_avec:.8f}")
print("-> sans residuel, le gradient meurt ; avec, il reste vivant")
```

```python
# --- BLOC 2 : construire et tester un bloc encodeur complet ---
from transformers_arch.encoder import EncoderBlock, TransformerEncoder

torch.manual_seed(42)
bloc = EncoderBlock(d_model=8, n_heads=2, d_ff=32)
phrase = torch.randn(1, 4, 8)

sortie, poids = bloc(phrase)
print("\nForme entree :", phrase.shape)
print("Forme sortie :", sortie.shape, "-> identique, donc empilable")
print("Parametres du bloc :", sum(p.numel() for p in bloc.parameters()))
```

```python
# --- BLOC 3 : verifier la normalisation (LayerNorm) ---
entree_mal_calibree = torch.randn(1, 3, 16) * 100  # valeurs volontairement extremes
bloc2 = EncoderBlock(d_model=16, n_heads=4, d_ff=64)
sortie2, _ = bloc2(entree_mal_calibree)

print(
    "\nMoyenne de l'entree (avant normalisation) :", entree_mal_calibree.mean().item()
)
print(
    "Moyenne de la sortie (apres LayerNorm), par position :",
    sortie2.mean(dim=-1)[0].detach().numpy().round(6),
)
print("-> chaque position recentree proche de 0, malgre une entree extreme")
```

```python
# --- BLOC 4 : empiler plusieurs blocs (comme un vrai Transformer) ---
torch.manual_seed(42)
for n_layers in [1, 3, 6, 12]:
    encodeur = TransformerEncoder(d_model=8, n_heads=2, d_ff=32, n_layers=n_layers)
    nb_params = sum(p.numel() for p in encodeur.parameters())
    sortie_empilee = encodeur(phrase)
    print(
        f"n_layers={n_layers:3} -> {nb_params:6,} parametres, "
        f"sortie {tuple(sortie_empilee.shape)}"
    )
```

## 3.   Fine Tuning

```python
# Prerequis : uv add transformers torch
# Necessite un acces reseau a huggingface.co (bloque dans le sandbox
# utilise pour ecrire ce code, donc a lancer chez toi -- meme situation
# que sentence-transformers en Phase 2 et spaCy en Phase 3)
# Utilise les fonctions reelles de src/transformer_arch/fine_tuning.py

# --- BLOC 1 : charger DistilBERT pre-entraine ---
import os
import sys

sys.path.insert(0, os.path.abspath(os.path.join("..", "src")))
from transformers_arch.fine_tuning import (
    SentimentDataset,
    evaluate_fine_tuned_model,
    fine_tune_model,
    load_pretrained_classifier,
    push_to_hub,
    tokenize_dataset,
)

model, tokenizer = load_pretrained_classifier("distilbert-base-uncased")
print("Nombre de parametres du modele :", sum(p.numel() for p in model.parameters()))
# A verifier : environ 67 millions de parametres pour DistilBERT --
# bien plus que tout ce qu'on a construit en Phase 5 (le plus gros
# etait le CNN, ~270 000 parametres).
```

```python
# --- BLOC 2 : preparer le vrai dataset movie_reviews ---
from classical_ml.data_loader import load_movie_reviews

X_train, X_test, y_train, y_test = load_movie_reviews()
y_train_num = [1 if lab == "pos" else 0 for lab in y_train]
y_test_num = [1 if lab == "pos" else 0 for lab in y_test]

# max_length=256 par defaut tronque 98% des avis (moyenne 746 mots) --
# DistilBERT limite a 512 tokens maximum (contrainte du modele) ;
# on teste ici avec le maximum autorise pour limiter la perte
# d'information, meme si ca ne couvre toujours pas l'integralite
train_encodings = tokenize_dataset(X_train, tokenizer, max_length=512)
test_encodings = tokenize_dataset(X_test, tokenizer, max_length=512)

train_dataset = SentimentDataset(train_encodings, y_train_num)
test_dataset = SentimentDataset(test_encodings, y_test_num)

print("Taille du train :", len(train_dataset))
print("Taille du test  :", len(test_dataset))
```

```python
trainer = fine_tune_model(
    model, train_dataset, test_dataset, epochs=5, learning_rate=2e-5
)
resultats = evaluate_fine_tuned_model(trainer)
print("\nResultats d'evaluation :", resultats)
```

```python
# --- BLOC 4 : comparaison avec TOUT ce qu'on a construit avant ---
print("\n{:35} {:>10}".format("Modele", "Accuracy"))
print("{:35} {:>10.3f}".format("Linear SVM (Phase 4)", 0.835))
print("{:35} {:>10.3f}".format("CNN + GloVe (Phase 5)", 0.630))
print(
    "{:35} {:>10.3f}".format(
        "DistilBERT fine-tune (Phase 6)", resultats.get("eval_accuracy", float("nan"))
    )
)
# A verifier chez toi : DistilBERT devrait nettement depasser le SVM,
# confirmant l'hypothese posee a la fin de la Phase 5 -- avec assez de
# connaissance pre-entrainee, les architectures profondes deviennent
# enfin superieures aux methodes classiques, meme sur seulement 1600
# exemples specifiques a la tache.
```
